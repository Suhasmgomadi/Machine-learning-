```python
def extract_pdf_text(file_path):

    text = ""

    document = fitz.open(file_path)

    for page in document:
        text += page.get_text()

    document.close()

    return text
```

```python
def extract_skills(text):
    found_skills = []
    text_lower = text.lower()
    for skill in SKILLS:
        # Use a regex to find the skill as a whole word
        if re.search(r'\b' + re.escape(skill) + r'\b', text_lower):
            found_skills.append(skill)
    return found_skills

resume_skills = extract_skills(resume_text)

print("Detected Skills:")
for skill in resume_skills:
    print("✓", skill)
```

```python
DEGREE_PATTERNS = {

    "B.Tech": [
        r"\bb\.?\s*tech\b",
        r"\bbtech\b",
        r"\bbachelor of technology\b"
    ],

    "B.E.": [
        r"\bb\.?\s*e\.?\b",
        r"\bbe\b",
        r"\bbachelor of engineering\b"
    ],

    "M.Tech": [
        r"\bm\.?\s*tech\b",
        r"\bmtech\b",
        r"\bmaster of technology\b"
    ],

    "M.E.": [
        r"\bm\.?\s*e\.?\b",
        r"\bme\b",
        r"\bmaster of engineering\b"
    ],

    "BCA": [
        r"\bbca\b",
        r"\bbachelor of computer applications\b"
    ],

    "MCA": [
        r"\bmca\b",
        r"\bmaster of computer applications\b"
    ],

    "B.Sc": [
        r"\bb\.?\s*sc\b",
        r"\bbachelor of science\b"
    ],

    "M.Sc": [
        r"\bm\.?\s*sc\b",
        r"\bmaster of science\b"
    ],

    "MBA": [
        r"\bmba\b",
        r"\bmaster of business administration\b"
    ],

    "PhD": [
        r"\bph\.?\s*d\b",
        r"\bdoctor of philosophy\b"
    ]
}
```

```python
def extract_degrees(text):

    text = text.lower()

    degrees = []

    for degree, patterns in DEGREE_PATTERNS.items():

        for pattern in patterns:

            if re.search(pattern, text):
                degrees.append(degree)
                break

    return sorted(set(degrees))
```

```python
BRANCHES = [
    "computer science",
    "computer science engineering",
    "information technology",
    "information science",
    "artificial intelligence",
    "artificial intelligence and machine learning",
    "data science",
    "electronics",
    "electronics and communication",
    "electrical engineering",
    "mechanical engineering",
    "civil engineering",
    "automobile engineering",
    "mechatronics",
    "chemical engineering"
]
```

```python
def extract_branches(text):

    text = text.lower()

    branches = []

    for branch in BRANCHES:

        if branch in text:
            branches.append(branch)

    return sorted(set(branches))
```

```python
def extract_experience(text):

    text = text.lower()

    patterns = [
        r"(\d+)\+?\s*years?\s*(?:of)?\s*experience",
        r"(\d+)\+?\s*years?\s*work experience",
        r"experience\s*[:\-]?\s*(\d+)\+?\s*years?"
    ]

    values = []

    for pattern in patterns:

        matches = re.findall(
            pattern,
            text
        )

        for value in matches:

            try:
                values.append(int(value))
            except:
                pass

    if values:
        return max(values)

    return 0
```

```python
def identify_profile(skills):

    skill_set = set(
        skill.lower()
        for skill in skills
    )

    if {"machine learning", "python"}.issubset(skill_set):
        return "Machine Learning / AI"

    if {"data science", "python"}.issubset(skill_set):
        return "Data Science"

    if {"nlp", "python"}.issubset(skill_set):
        return "NLP / AI"

    if {"react", "javascript"}.issubset(skill_set):
        return "Frontend Development"

    if {"django", "python"}.issubset(skill_set):
        return "Backend Development"

    if {"sql", "power bi"}.issubset(skill_set):
        return "Data Analytics"

    if {"c++", "embedded systems"}.issubset(skill_set):
        return "Embedded Systems"

    return "Software / Technology"
```

```python
def calculate_skill_match(resume_skills, job_skills):

    resume_set = set(
        str(skill).lower().strip()
        for skill in resume_skills
    )

    job_set = set(
        skill.strip().lower()
        for skill in str(job_skills).split(",")
        if skill.strip()
    )

    if len(job_set) == 0:
        return 0, [], []

    matched = sorted(
        resume_set.intersection(job_set)
    )

    missing = sorted(
        job_set.difference(resume_set)
    )

    score = (
        len(matched) / len(job_set)
    ) * 100

    return score, matched, missing
```

```python
def calculate_education_match(
    resume_degrees,
    job_education
):

    if not resume_degrees:
        return 50

    job_education = str(
        job_education
    ).lower()

    if "any" in job_education:
        return 100

    for degree in resume_degrees:

        degree = degree.lower()

        if degree in job_education:
            return 100

        # Flexible matching
        if degree == "b.tech" and (
            "btech" in job_education or
            "b.tech" in job_education
        ):
            return 100

        if degree == "m.tech" and (
            "mtech" in job_education or
            "m.tech" in job_education
        ):
            return 100

    return 40
```

```python
def calculate_experience_match(
    resume_experience,
    minimum_experience,
    maximum_experience
):

    try:
        resume_experience = float(
            resume_experience
        )

        minimum_experience = float(
            minimum_experience
        )

        maximum_experience = float(
            maximum_experience
        )

    except:
        return 50

    # Perfect range
    if (
        minimum_experience
        <= resume_experience
        <= maximum_experience
    ):
        return 100

    # Candidate has slightly less experience
    if resume_experience < minimum_experience:

        difference = (
            minimum_experience
            - resume_experience
        )

        if difference <= 1:
            return 75

        if difference <= 2:
            return 50

        return 25

    # Candidate has more experience
    if resume_experience > maximum_experience:
        return 90

    return 50
```

```python
def generate_job_recommendations(
    resume_text,
    top_n=10
):

    # -----------------------------
    # Analyze resume
    # -----------------------------

    resume_skills = extract_skills(
        resume_text
    )

    resume_degrees = extract_degrees(
        resume_text
    )

    resume_branches = extract_branches(
        resume_text
    )

    resume_experience = extract_experience(
        resume_text
    )

    resume_profile = identify_profile(
        resume_skills
    )

    # -----------------------------
    # TF-IDF similarity
    # -----------------------------

    resume_vector = tfidf.transform(
        [resume_text]
    )

    similarities = cosine_similarity(
        resume_vector,
        job_vectors
    ).flatten()

    results = []

    # -----------------------------
    # Score every job
    # -----------------------------

    for i, job in jobs_df.iterrows():

        tfidf_score = (
            similarities[i] * 100
        )

        skill_score, matched_skills, missing_skills = (
            calculate_skill_match(
                resume_skills,
                job["skills"]
            )
        )

        education_score = (
            calculate_education_match(
                resume_degrees,
                job["education"]
            )
        )

        experience_score = (
            calculate_experience_match(
                resume_experience,
                job["experience_min"],
                job["experience_max"]
            )
        )

        # -----------------------------
        # Final weighted score
        # -----------------------------

        final_score = (
            0.50 * tfidf_score +
            0.30 * skill_score +
            0.10 * education_score +
            0.10 * experience_score
        )

        results.append({

            "Job Title":
                job["job_title"],

            "Company":
                job["company"],

            "Source":
                job["source"],

            "Location":
                job["location"],

            "Education":
                job["education"],

            "Experience":
                str(job["experience_min"])
                + " - "
                + str(job["experience_max"])
                + " years",

            "Match Score":
                round(final_score, 2),

            "TF-IDF Score":
                round(tfidf_score, 2),

            "Skill Score":
                round(skill_score, 2),

            "Matched Skills":
                ", ".join(matched_skills),

            "Missing Skills":
                ", ".join(missing_skills),

            "Job URL":
                job["job_url"]

        })

    results_df = pd.DataFrame(
        results
    )

    results_df = results_df.sort_values(
        "Match Score",
        ascending=False
    )

    return {
        "profile": resume_profile,
        "skills": resume_skills,
        "degrees": resume_degrees,
        "branches": resume_branches,
        "experience": resume_experience,
        "recommendations": results_df.head(top_n)
    }
```

```python
def process_resume(file):

    if file is None:

        return (
            "Please upload a resume.",
            "",
            "",
            "",
            "",
            "",
            pd.DataFrame()
        )

    try:

        file_path = file.name

        # -------------------------
        # Extract text
        # -------------------------

        if file_path.lower().endswith(".pdf"):

            text = extract_pdf_text(
                file_path
            )

        elif file_path.lower().endswith(".docx"):

            text = extract_docx_text(
                file_path
            )

        elif file_path.lower().endswith(".txt"):

            with open(
                file_path,
                "r",
                encoding="utf-8"
            ) as f:

                text = f.read()

        else:

            return (
                "Unsupported file type.",
                "",
                "",
                "",
                "",
                "",
                pd.DataFrame()
            )

        # -------------------------
        # Run recommendation engine
        # -------------------------

        results = generate_job_recommendations(
            text,
            top_n=10
        )

        profile = results["profile"]

        skills = ", ".join(
            results["skills"]
        )

        degrees = ", ".join(
            results["degrees"]
        )

        branches = ", ".join(
            results["branches"]
        )

        experience = (
            str(results["experience"])
            + " years"
        )

        recommendations = (
            results["recommendations"]
        )

        return (
            "Resume analyzed successfully!",
            profile,
            skills,
            degrees,
            branches,
            experience,
            recommendations
        )

    except Exception as e:

        return (
            "Error: " + str(e),
            "",
            "",
            "",
            "",
            "",
            pd.DataFrame()
        )
```
